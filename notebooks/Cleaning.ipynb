```python
campaigns['start_date']=pd.to_datetime(campaigns['start_date'])
campaigns['end_date']=pd.to_datetime(campaigns['end_date'])
```

```python
transactions.info()
transactions.isnull().sum()
```

```python
transactions['timestamp']=pd.to_datetime(transactions['timestamp'])
```

```python
transactions = transactions.dropna(subset=['product_id'])
```

```python
transactions = transactions.dropna(subset=['product_id'])
```

```python
transactions['gross_revenue'] = transactions['gross_revenue'].fillna(0)
```

```python
transactions['product_id'] = transactions['product_id'].astype(int)
transactions['campaign_id'] = transactions['campaign_id'].astype(int)
```

```python
transactions[transactions.duplicated()]
```

```python
transactions['net_revenue'] = transactions['gross_revenue'] - transactions['discount_applied']
transactions['revenue_per_item'] = transactions['gross_revenue'] / transactions['quantity']
```

```python
events.head()
```

```python
events.info()
events.isnull().sum()
```

```python
events[events.duplicated()]
```

```python
events['timestamp'] = pd.to_datetime(events['timestamp'])
```

```python
events['event_type'] = events['event_type'].str.strip().str.lower()
events['traffic_source'] = events['traffic_source'].str.strip().str.title()
events['device_type'] = events['device_type'].str.strip().str.title()
events['experiment_group'] = events['experiment_group'].str.strip().str.title()
events['page_category'] = events['page_category'].str.strip().str.lower()
```

```python
events['product_id'] = events['product_id'].fillna(0)
```

```python
events['device_type'] = events['device_type'].fillna("Unknown")
```

```python
events['product_id'] = events['product_id'].astype(int)
events['campaign_id'] = events['campaign_id'].astype(int)
```

```python
events['is_purchase'] = (events['event_type'] == 'purchase').astype(int)
```

```python
events['hour'] = events['timestamp'].dt.hour
events['day'] = events['timestamp'].dt.day_name()
```

```python
events['event_type'].value_counts()
```

```python
events.groupby('experiment_group')['is_purchase'].mean()
events.groupby('device_type')['is_purchase'].mean()
```
